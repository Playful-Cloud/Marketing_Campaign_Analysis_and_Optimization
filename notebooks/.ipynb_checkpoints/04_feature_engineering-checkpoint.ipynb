# Features Engineering:
---

Feature engineering was performed to create additional marketing performance metrics and time-based insights, enabling more accurate analysis and future machine learning experimentation.

### Objectives:
- Create new analytical marketing KPIs.
- Generate time-based campaign performance attributes.
- Prepare enriched dataset for modeling and deeper insight extraction.

### Engineered Features:

---
| Feature                             | Formula                                 | Meaning                             |
| ----------------------------------- | --------------------------------------- | ----------------------------------------------------- |
| CTR (Click-Through Rate)            | `clicks / impressions`                  | Measures audience engagement with ads                 |
| Conversion Rate (CR)                | `conversions / clicks`                  | Measures how efficiently clicks turn into conversions |
| CPC (Cost per Click)                | `spend_usd / clicks`                    | Shows cost efficiency of attracting each click        |
| CPM (Cost per Thousand Impressions) | `spend_usd / (impressions / 1000)`      | Shows cost to reach 1k people                         |
| CPA (Cost per Acquisition)          | `spend_usd / conversions`               | Cost per conversion/ critical efficiency metric      |
| ROI (Return on Investment)          | `(revenue_usd - spend_usd) / spend_usd` | Measures profitability of marketing spend             |
| ROAS (Return on Ad Spend)           | `revenue_usd / spend_usd`               | Shows revenue return per dollar spent                 |
| Revenue per Click                   | `revenue_usd / clicks`                  | How much each click generates in revenue              |
| Revenue per Conversion              | `revenue_usd / conversions`             | Revenue per successful conversion                     |
| Campaign Duration (days)            | `end_date - start_date`                 | Campaign timeline context                             |


---


#### Features (Extended):

---

| Feature                     | Formula                                           | Meaning                                                                 |
|-----------------------------|----------------------------------------------------|-------------------------------------------------------------------------|
| Profit                      | `revenue_usd - spend_usd`                          | Total net earnings after subtracting marketing spend.                   |
| Profit Margin               | `profit / revenue_usd`                             | Percentage of revenue retained as profit.                               |
| Impressions to Conversion   | `impressions / conversions`                        | Number of impressions needed to generate one conversion.                |
| Click-to-Conversion Ratio   | `clicks / conversions`                             | Number of clicks required to produce one conversion.                    |
| Midpoint Month              | `start_date + (campaign_duration_days / 2)` `.month` | Central month of campaign; used to detect seasonality patterns.         |
| Engagement Rate per 1000    | `(clicks / impressions) * 1000`                    | Clicks per 1000 impressions; normalized engagement metric.              |
| Revenue per 1000 (RPM)      | `revenue_usd / (impressions / 1000)`               | Revenue generated per 1000 ad impressions.                              |
---

### Data Handling Notes:

- Zeros in clicks, impressions, conversions, spend_usd are temporarily treated as NaN to avoid invalid division errors.

- After metric calculation, NaN values are safely converted back to zero for reporting clarity.

### Purpose:

These engineered features form the foundation for:
- Performance benchmarking across campaigns and channels
- Exploratory data analysis (EDA)
- Predictive modeling (ROI forecasting, conversion modeling)
- Automation workflows in the MLOps pipeline (future phase)

---

## Convert Data types - Dates, Float & Numeric:

## Features Engineering (CORE):

```python
# ADD CUSTOM COLUMNS: 
# ctr
# conversion_rate 
# roi 
# campaign_duration_days


# # Performance Metrics
# df['ctr'] = np.where(df['impressions'] > 0, df['clicks'] / df['impressions'], 0)
# df['conversion_rate'] = np.where(df['clicks'] > 0, df['conversions'] / df['clicks'], 0)
# df['roi'] = np.where(df['spend_usd'] > 0, (df['revenue_usd'] - df['spend_usd']) / df['spend_usd'], 0)
# df['roas'] = np.where(df['spend_usd'] > 0, df['revenue_usd'] / df['spend_usd'], 0)

# # Cost Metrics 
# df['cpc'] = np.where(df['clicks'] > 0, df['spend_usd'] / df['clicks'], 0)
# df['cpm'] = np.where(df['impressions'] > 0, df['spend_usd'] / (df['impressions'] / 1000), 0)
# df['cpa'] = np.where(df['conversions'] > 0, df['spend_usd'] / df['conversions'], 0)

# # Revenue Efficiency Metrics
# df['revenue_per_click'] = np.where(df['clicks'] > 0, df['revenue_usd'] / df['clicks'], 0)
# df['revenue_per_conversion'] = np.where(df['conversions'] > 0, df['revenue_usd'] / df['conversions'], 0)
# #updated df


# # Rounding

# df[['ctr', 'conversion_rate', 'roi', 'roas', 'cpc', 'cpm', 'cpa',
#     'revenue_per_click', 'revenue_per_conversion']] = df[[
#     'ctr', 'conversion_rate', 'roi', 'roas', 'cpc', 'cpm', 'cpa',
#     'revenue_per_click', 'revenue_per_conversion'
# ]].round(6)



# Performance Metrics
df['ctr'] = np.where(df['impressions'] > 0, df['clicks'] / df['impressions'], 0)
df['conversion_rate'] = np.where(df['clicks'] > 0, df['conversions'] / df['clicks'], 0)
df['roi'] = np.where(df['spend_usd'] > 0, (df['revenue_usd'] - df['spend_usd']) / df['spend_usd'], 0)
df['roas'] = np.where(df['spend_usd'] > 0, df['revenue_usd'] / df['spend_usd'], 0)

# Cost Metrics
df['cpc'] = np.where(df['clicks'] > 0, df['spend_usd'] / df['clicks'], 0)
df['cpm'] = np.where(df['impressions'] > 0, df['spend_usd'] / (df['impressions'] / 1000), 0)
df['cpa'] = np.where(df['conversions'] > 0, df['spend_usd'] / df['conversions'], 0)

# Revenue Efficiency
df['revenue_per_click'] = np.where(df['clicks'] > 0, df['revenue_usd'] / df['clicks'], 0)
df['revenue_per_conversion'] = np.where(df['conversions'] > 0, df['revenue_usd'] / df['conversions'], 0)

# Rounding core metrics
df[['ctr', 'conversion_rate', 'roi', 'roas', 'cpc', 'cpm', 'cpa',
    'revenue_per_click', 'revenue_per_conversion']] = df[[
    'ctr', 'conversion_rate', 'roi', 'roas', 'cpc', 'cpm', 'cpa',
    'revenue_per_click', 'revenue_per_conversion'
]].round(6)
```

## Sanity Checks / Quick Validations:

#### Sanity Validation Columns:

```python
# my Sanity validation check columns

print("\nFEATURE ENGINEERING CONSISTENCY CHECK")

# Check ROAS consistency
df['roas_check'] = df['revenue_usd'] / df['spend_usd']
roas_mismatch = (df['roas'] - df['roas_check']).abs() > 1e-6
print(f"ROAS mismatches: {roas_mismatch.sum()}")

# Check ROI consistency (ROI = (revenue - spend) / spend)
df['roi_check'] = (df['revenue_usd'] - df['spend_usd']) / df['spend_usd']
roi_mismatch = (df['roi'] - df['roi_check']).abs() > 1e-6
print(f"ROI mismatches: {roi_mismatch.sum()}")

# Check CPC consistency (spend/clicks)
df['cpc_check'] = df['spend_usd'] / df['clicks']
cpc_mismatch = (df['cpc'] - df['cpc_check']).abs() > 1e-6
print(f"CPC mismatches: {cpc_mismatch.sum()}")

# Check CPM consistency (spend / (impressions/1000))
df['cpm_check'] = df['spend_usd'] / (df['impressions'] / 1000)
cpm_mismatch = (df['cpm'] - df['cpm_check']).abs() > 1e-6
print(f"CPM mismatches: {cpm_mismatch.sum()}")

# Check CPA consistency (spend/conversions)
df['cpa_check'] = df['spend_usd'] / df['conversions']
cpa_mismatch = (df['cpa'] - df['cpa_check']).abs() > 1e-6
print(f"CPA mismatches: {cpa_mismatch.sum()}")

# Check CTR consistency
df['ctr_check'] = df['clicks'] / df['impressions']
ctr_mismatch = (df['ctr'] - df['ctr_check']).abs() > 1e-6
print(f"CTR mismatches: {ctr_mismatch.sum()}")

# Check Conversion Rate consistency
df['conversion_rate_check'] = df['conversions'] / df['clicks']
cr_mismatch = (df['conversion_rate'] - df['conversion_rate_check']).abs() > 1e-6
print(f"Conversion Rate mismatches: {cr_mismatch.sum()}")

print("CONSISTENCY CHECK COMPLETE !!!")
```

```python
# Drop the Sanity validation check columns

df.drop(columns=[
    'roas_check', 'roi_check', 'cpc_check', 'cpm_check',
    'cpa_check', 'ctr_check', 'conversion_rate_check'
], inplace=True)
```

```python

# Click-to-Conversion Ratio
df['click_to_conv_ratio'] = df['conversions'] / df['clicks']

# Mid-Campaign Month
df['midpoint_date'] = df['start_date'] + pd.to_timedelta(df['campaign_duration_days'] / 2, unit='D')
df['midpoint_month'] = df['midpoint_date'].dt.month

# Season Label from Quarter
season_map = {1: 'New Year', 2: 'Summer', 3: 'Back-to-school', 4: 'Holiday Peak'}
df['season_label'] = df['quarter'].map(season_map)

# Engagement per 1,000 impressions
df['engagement_rate_1000'] = (df['clicks'] + df['conversions']) / (df['impressions'] / 1000)

# Revenue per 1,000 impressions
df['rev_per_1000'] = df['revenue_usd'] / (df['impressions'] / 1000)

# ROI Flag (optional for ML)
roi_median = df['roi'].median()
df['roi_flag'] = (df['roi'] > roi_median).astype(int)
```

### Correlation Matrix - Relationships Between Metrics:

The correlation matrix reveals how the engineered performance metrics relate to one another. This helps ensure that derived values behave logically and that redundant or misleading relationships can be flagged before modeling.

#### Correlation Matrix:

#### My Colorful HeatMap:

```python
# Relationships Between Metrices HEATMAP

numeric_features = ['ctr', 'conversion_rate', 'roi', 'roas', 'cpc', 'cpm', 'cpa', 
                    'revenue_per_click', 'revenue_per_conversion']

corr = df[numeric_features].corr()
plt.figure(figsize=(10, 6))
# plt.imshow(corr, cmap='coolwarm', interpolation='nearest')
# # plt.colorbar(label='Correlation')
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap of Engineered Features")
# plt.tight_layout()
plt.show()
```

##### Key Observations:

- ##### Strong Positive Correlations (Values >= 0.75):
    - ROI and ROAS (1.00) There is a perfect positive correlation between return on investment (ROI) and return on advertising spend (ROAS). This is expected because both metrics measure profitability using almost identical formulas. They will always move together.
    
    - CPA and Revenue per Conversion (0.99) These two metrics show an extremely strong correlation. This indicates that conversions that cost more to acquire tend to generate more revenue per conversion. This is common in tiered or value-based acquisition environments.

- ##### Moderate Positive Correlations (Values between 0.4 and 0.75):
    - CPC and Revenue per Click (0.74) Click costs correlate strongly with revenue generated per click. Higher-quality or more competitive traffic often comes with higher CPCs, but also tends to be more valuable.
      
    - CTR and Conversion Rate (0.60) Campaigns that get more people to click also tend to convert better. This indicates aligned messaging or targeting effectiveness across funnel stages.

    - Revenue per Click and Revenue per Conversion (0.56) Not as strong as the above, but still notable. This suggests that campaigns generating high value per click also tend to produce high value per conversion.
 
    - CPA and Revenue per Click (0.55) Acquisitions that cost more correlate with clicks that generate more revenue, reinforcing the idea that high-value conversions require higher investment.
 
    - CPC and CPA (0.46) As the cost per click increases, the cost to acquire a customer tends to rise as well. This reflects how competitive or high-value audiences behave.

    - CPC and Revenue per Conversion (0.41) Higher click prices often correspond to higher conversion value, implying that high-value audiences are expensive to reach.
 
- ##### Weak or Insignificant Correlations (Values near 0.0):
    - CPM and CPA (0.05) Exposure cost has almost no connection with acquisition cost. This is common, as CPM reflects top-of-funnel reach while CPA reflects bottom-of-funnel efficiency.

    - CPM and Revenue per Conversion (0.05) Again, very weak. Cost to show ads rarely predicts the revenue those conversions will generate.

- ##### Negative Correlations (Blue cells):
    - ROI/ROAS and CPA (–0.04) Nearly no relationship between cost per acquisition and profitability metrics. This means profitability is likely being driven more by revenue efficiency rather than cost efficiency.
      
    - Conversion Rate and CPC (–0.08) A slight negative relationship suggesting that higher click costs do not necessarily translate to higher conversion efficiency.

    - Conversion Rate and Revenue per Click (–0.09) Weak inverse relationship, indicating that valuable clicks may not always convert at higher rates.
    
    - Conversion Rate and Revenue per Conversion (–0.16) Slightly stronger but still weak. High-value conversions can come at the cost of lower conversion efficiency.
    
    - CTR and CPC (–0.19) Higher click costs sometimes correlate with lower click-through rates, which can happen when ads target niche or competitive audiences.

##### Remarks:

- Most relationships appear logical and consistent with marketing economics. Higher costs tend to reflect higher-value audiences, and revenue-based metrics align cleanly with cost-based metrics.

- No unexpected inverse correlations were found.
The few negative relationships are weak and represent noise or expected behavioral variance.

- ROI and ROAS are fully redundant. These metrics have a perfect correlation and should not be used together in modeling to avoid multicollinearity.

- Some cost metrics may require regularization. CPC, CPA, and revenue-per-click-related metrics show moderate-to-strong interdependence, which can introduce variance inflation in regression models.

- Conversion metrics remain relatively independent. CTR and conversion rate maintain moderate connection but not excessive overlap, making them useful as separate explanatory variables.

#### Campaign duration Validation: ROI & ROAS by Duration Range (Grouped summary)

This is a grouped summary validation version of the earlier scatter plot, now using boxplots to compare ROI and ROAS distributions by campaign duration range.

```python
# ROI & ROAS by Duration Range (Grouped summary)

# Define duration bins and labels
bins = [0, 30, 90, 180, 365]
labels = ['<1 mo', '1–3 mo', '3–6 mo', '6–12 mo']
df['duration_group'] = pd.cut(df['campaign_duration_days'], bins=bins, labels=labels, include_lowest=True)

# Boxplots for ROI and ROAS across duration groups
plt.figure(figsize=(12, 6))
for i, col in enumerate(['roi', 'roas'], 1):
    plt.subplot(1, 2, i)
    sns.boxplot(x='duration_group', y=col, data=df)
    plt.title(f"{col.upper()} by Campaign Duration Group")
    plt.xlabel("Campaign Duration Group")
    plt.ylabel(col.upper())

    
plt.tight_layout()
plt.show()

# Group and aggregate mean ROI/ROAS
# duration_summary = df.groupby('duration_group')[['roi', 'roas']].mean().reset_index()
duration_summary = (
    df.groupby('duration_group', observed=True)[['roi', 'roas']]
      .mean()
      .reset_index()
)

# Bar chart of average ROI and ROAS
plt.figure(figsize=(12, 6))
bar_width = 0.35

plt.bar(duration_summary['duration_group'], duration_summary['roas'], width=bar_width, label='ROAS', alpha=0.7)
plt.bar(duration_summary['duration_group'], duration_summary['roi'], width=bar_width, label='ROI', alpha=0.7)

plt.title("Average ROI and ROAS by Campaign Duration Range")
plt.xlabel("Campaign Duration Range")
plt.ylabel("Average Value")
plt.legend()


plt.tight_layout()
plt.show()
```

##### Key Observations:

- This section summarizes how ROI and ROAS vary across campaign duration groups, combining boxplots (for spread and outliers) and grouped averages (for trend visibility).
- It builds on the previous duration validation to reveal whether longer campaigns deliver better returns or simply exhibit higher variability.
- it apears that ROI and ROAS fluctuate meaningfully across duration groups, & it confirms that performance behavior changes with campaign length.
- This distribution pattern supports treating duration as a nonlinear or interaction feature in predictive models, not as a purely linear driver.
- ROI and ROAS box plots, the median line (the line inside the blue box) is consistently near zero across all duration groups. There is no significant increase in the typical (median) performance between the 1-3 month group and the others.
- The average increases with duration, but the typical performance for a given campaign remains low regardless of its length (meaning, high average is driven by outliers in longer campaigns, not a higher median performance)

    - Use RobustScaler() or StandardScaler() to normalize numeric features, ensuring comparability and model stability under varying data scales.
- Encoding & Interaction Enrichment:
    - Apply one-hot encoding for key categorical variables (Channel, Region, Duration Group) to preserve interpretability.
    - Engineer interaction features such as Channel × Duration and Region × ROI to capture cross-market and contextual performance effects
- Modeling Direction:
    - Begin with ElasticNet Regression to identify linear feature relationships and manage regularization between correlated predictors.
    - Follow with RandomForestRegressor to capture non-linear and interaction-driven effects, while providing interpretable feature importance insights.
    - Validate results via cross-validation to confirm consistency and generalizability across campaign types and durations.
